## 🔹 Python vs C Arrays (Memory Model)

* **C/Java Arrays** → Store the **raw data values** in contiguous memory.

  * Example: `int arr[5]` → just raw integers, each 4 bytes, packed one after the other.
  * Restriction: only **one data type**.
  * Advantage: **fast, compact**.

* **Python Lists** → Store **references (pointers) to Python objects** in contiguous memory.

  * Example: `arr = [1, "hello", 3.5]` → list stores **pointers** to integer object `1`, string object `"hello"`, float object `3.5`.
  * Advantage: allows **mixed data types**.
  * Disadvantage: extra indirection → **slower & memory-heavier**.

---

## 🔹 Limitations of Arrays (final list)

1. **Homogeneous restriction (in low-level arrays)**

   * C/Java arrays can only hold one data type.
   * Python lists overcome this, but at the cost of performance.

2. **Insertion/Deletion Cost**

   * At beginning/middle → need to shift elements → **O(n)**.
   * At end (append, pop) → **O(1)** amortized.

3. **Fixed Size (in low-level arrays)**

   * C arrays cannot grow; you must reallocate.
   * Python lists resize automatically, but resizing requires copying all elements → **O(n)** in worst case.

4. **Memory Overhead (Python-specific)**

   * Python stores references + object metadata.
   * So Python lists use **much more memory** than raw arrays in C.

5. **Cache Performance**

   * Low-level arrays = very cache-friendly.
   * Python lists = less cache-friendly because of pointers + scattered object data.

## ✅ Review Answer (how you should phrase it)

> “Arrays are stored in contiguous memory, which allows O(1) random access. But they have limitations: in low-level languages, they are fixed size and store only one type. Insertion and deletion in the middle or beginning is costly — O(n). In Python, the list type overcomes size and type restrictions by storing references, but this adds memory overhead and makes operations slower compared to C arrays.”

🔹 2. Indexing & Random Access

Indexing starts at 0.

Each element accessed directly with O(1) time.

Example:

```python
arr = [10,20, 30, 40, 50]

print(arr[0]) #10 (first element)
print(arr[3]) #40 fourth element
print(arr[-1]) #50 (last element, negative indexing allowed)
# ⚡ Big-O for access: O(1).
```

1) Insert
a) Insert at end (append) — amortized O(1)

```python
arr = [10,20,30]
arr.append(40) # O(1) 
```

```python
# b) Insert at index i — O(n) (shift right)

def insert_at(arr, i, val):
    if i < 0:
        i = 0
    if i > len(arr):
        i = len(arr)
    arr.insert(i, val)
    return arr
# Demo
a = [10, 20, 30, 40]
insert_at(a, 2, 99)        
```

```python
a = [10,20,30,40]
insert_at(a,2,99)
```

```python
arr = [10, 20, 30]
x = arr.pop()

#a) Delete from end — O(1)
```

```python
def delete_at(arr, i):
    if 0 <= i < len(arr):
        arr.pop(i)
    return arr

a = [10,20,30,40]
delete_at(a, 1)
# Delete at index i — O(n) (shift left)
```

```python
def delete_value(arr, val):
    for i, x in enumerate(arr):
        if x == val:
            arr.pop(i)
            break
    return arr

# Delete by value (first match) — O(n) to find + shift
```

```python
# 3) Update (set value at index) — O(1)

arr = [5, 6, 7]
arr[1] = 99          # [5, 99, 7]
```

```python
# a) Linear Search — O(n) (works on any array)
def linaer_search(arr, target):
    for i, x in enumerate(arr):
        if x == target:
            return i
    return -1

linaer_search([10,20,30], 20)
linaer_search([10,20, 30], 50)
```

```python
arr.reverse()              # in place
# or
arr = arr[::-1]            # new list (O(n) memory)
```

Rotate right by k — O(n)

```python
def rotate_right(arr, k):
    n = len(arr)
    k %= n
    arr[:] = arr[-k:] + arr[:-k]
```

```python
# Remove duplicates from sorted array — O(n)
def dedup_sorted(arr):
    if not arr:
        return arr
    write = 1
    for read in range(1, len(arr)):
        if arr[read] != arr[read-1]:
            arr[write] = arr[read]
            write += 1
    return arr[:write]
```
