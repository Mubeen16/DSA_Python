````python
def dedup_sorted(arr):
    """
    Remove duplicates from a sorted array in-place.
    Returns the new length (count of unique elements).
    After the function, arr[:new_len] contains all uniques.
    """
    n = len(arr)
    if n == 0:
        return 0
    w = 1
    for r in range(1,n):
        if arr[r] != arr[w-1]:
            arr[w]=arr[r]
            w += 1
    return w
````

````python
#Practice 1
def dup(arr):
    n = len(arr)
    if n == 0:
        return 0
    w = 1
    for r in range(1,n):
        if arr[r] != arr[w]:
            arr[w] = arr[r]
            w += 1
    return w
````

````python
def dedup_sorted_arr(arr):
    unique = sorted(set(arr))
    for i in range(len(unique)):
        arr[i] = unique[i]
        return len(unique)

'''
✅ Super short, works for any input.

❌ Uses extra space (set + new list), not allowed by question.

❌ Sorting again = O(n log n) (though the input is already sorted).
'''
````

````python
def dedup_sorted_arr(arr):
    unique = sorted(set(arr))
    for i in range(len(unique)):
        arr[i] = unique[i]
        return len(unique)
````

````python
def dedup_unsorted(arr):
    seen = set()
    w = 0
    for x in arr:
        if x not in seen:
            seen.add(x)
            arr[w] = x
            w += 1
    return w

'''
✅ Works for unsorted arrays too.

❌ Uses O(n) extra space.
'''
````

````python
def two_sum(arr, target):
    """
    Return indices [i, j] such that arr[i] + arr[j] == target.
    If no solution, return None.
    """
    seen = {}
    for i, num in enumerate(arr):
        comp = target - num
        if comp in seen:
            return [seen[comp], i]
        seen[num] = i
    return None
````

````python
def two_sum_brue(arr, target):
    n = len(arr)
    for i in range(n):
        for j in range(i+1, n):
            if arr[i] + arr[j] == target:
                return [i, j]
    return None
        
````
