```python
def search_insert(nums, target):
    """
    Binary search to find target or insertion index.
    Time: O(log n), Space: O(1)
    """
    l, r = 0, len(nums)-1

    while l <= r:
        mid = (l + r)//2
        if nums[mid] == target:
            return mid
        elif nums[mid]<target:
            l = mid + 1
        else:
            r = mid - 1
```
