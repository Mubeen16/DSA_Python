## What category is “Move Zeroes”?

**Category:** **Array manipulation → two-pointers → stable compaction (in-place).**
It’s **not** searching and **not** sorting. You’re just **rearranging** elements while keeping the relative order of non-zeros.

---

## Basic array categories (use this map)

1. **Traversal & Aggregates (baby steps)**

   * Sum of array
   * Max/min element
   * Count evens/odds/positives/negatives

2. **Searching**

   * **Linear search** (find index of a target)
   * Count occurrences of a value

3. **In-place Edits**

   * Insert at index (shift right)
   * Delete at index (shift left)
   * Reverse an array (in place)

4. **Rearrangement (no sorting)**

   * **Move Zeroes** (stable compaction) ← *your question lives here*
   * Remove all occurrences of `val` (return new length)
   * Stable partition: move negatives to front (keep order)

5. **Two-Pointers (intro)**

   * Remove duplicates from **sorted** array
   * Merge two **sorted** arrays into one (new array)
   * Check if array is palindrome (two ends)

6. **Prefix sums (easy math)**

   * Running sum
   * Pivot index (left sum == right sum)

7. **Sorting (only after the above)**

   * Selection sort (simplest to code)
   * Insertion sort (good logic muscle)
   * Bubble sort (practice swaps)

---

```python
def sum_array(nums):
    total = 0.0
    for i in range(len(nums)):
        total += nums[i]
    return total
```

```python
def sum_array_n(nums, n):
    total_n = 0
    for i in range(n):
        total_n += nums[i]
    return total_n
```

```python
def sum_even(nums):
    total_even = 0
    for i in range(len(nums)):
        if i % 2 == 0:
            total_even += nums[i]
    return total_even
```

```python
def add_positive(nums):
    total = 0
    for i in nums:
        if i <= 0:
            continue
            total += i
    return total
```

```python
def sum_abs(nums):
    total = 0
    for i in nums:
        if i >= 0:
            total += i
    return total

```
