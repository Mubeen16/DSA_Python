# 2) Second Largest (distinct)
Task: Return the second largest distinct value; return None if it doesn’t exist. Signature: def second_largest(arr): ... Target: O(n) time, O(1) space; single pass preferred. Tests:

[10, 20, 5, 8] → 10
[7,7,7] → None
[-2, -1, -5] → -2


```python
def second_larg(arr):
    if len(arr) < 2:
        return None

    first = second = float('-inf')
    for x in arr:
        if x > first:
            second, first = first, x
        elif first > x > second:
            second = x
    return None if second == float('-inf') else second
```

```python
def second_larg_sort(arr):
    s = sorted(set(arr))
    return s[-2] if len(s) >= 2 else None
```

# 3) Rotate Right by k (in-place)

```python
def rotate_right(arr, k):
    n = len(arr)
    if n == 0:
        return arr
    k %= n
    return arr[-k:] + arr[:-k]
```
