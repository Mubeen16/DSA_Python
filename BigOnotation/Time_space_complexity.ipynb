

## **1️⃣ Understanding Time Complexity with Real Examples**

Think of **time complexity** as:

> *"How does the number of steps grow when the input size grows?"*

---

### **O(1) — Constant Time**

💡 Steps needed **do not depend** on the size of the input.

Example: Accessing an element in a list by index.

```python
arr = [5, 10, 15, 20, 25]
print(arr[2])  # Always 1 step, no matter if arr has 5 or 5 million elements
```

Even if the array is huge, index access takes the same time.

---

### **O(n) — Linear Time**

💡 Steps grow **proportionally** to the size of the input.

Example: Traversing a list.

```python
arr = [1, 2, 3, 4, 5]
for x in arr:
    print(x)  # Runs once for each element
```

If you double the size of the array, steps double.

---

### **O(n²) — Quadratic Time**

💡 Steps grow as the **square** of input size (nested loops).

Example: Printing all pairs.

```python
arr = [1, 2, 3]
for i in arr:
    for j in arr:
        print(i, j)  # Runs n * n times
```

If `n` = 3 → 9 steps
If `n` = 1000 → 1,000,000 steps 😬

---

### **O(log n) — Logarithmic Time**

💡 Steps grow **slowly** compared to the size of the input — happens in divide & conquer algorithms like **binary search**.

Example: Binary search in a sorted array.

```python
arr = [1, 3, 5, 7, 9, 11, 13]
target = 11
low, high = 0, len(arr) - 1

while low <= high:
    mid = (low + high) // 2
    if arr[mid] == target:
        print("Found at index", mid)
        break
    elif arr[mid] < target:
        low = mid + 1
    else:
        high = mid - 1
```

If you double the array size, steps only increase by **+1** or so, because each step halves the search space.

## **2️⃣ Basic Array Operations (from scratch)**

We’ll **not use built-in append/remove** — we’ll manipulate arrays manually.

---

### **🔹 Traversal**

```python
arr = [10, 20, 30, 40, 50]
for i in range(len(arr)):
    print(f"Index {i} → Value {arr[i]}")
```

⏱ Time complexity: **O(n)**

---

### **🔹 Insertion**

Let’s insert `99` at index `2` manually.

```python
arr = [10, 20, 30, 40, 50]
pos = 2
value = 99

arr.append(0)  # temporary space
for i in range(len(arr) - 1, pos, -1):
    arr[i] = arr[i - 1]  # shift right
arr[pos] = value

print(arr)  # [10, 20, 99, 30, 40, 50]
```

⏱ Time complexity: **O(n)** (worst case when inserting at the start)

---

### **🔹 Deletion**

Let’s delete the element at index `3`.

```python
arr = [10, 20, 30, 40, 50]
pos = 3

for i in range(pos, len(arr) - 1):
    arr[i] = arr[i + 1]  # shift left
arr.pop()  # remove last extra slot

print(arr)  # [10, 20, 30, 50]
```

⏱ Time complexity: **O(n)** (worst case when deleting from start)
