````python
def match_cookies(greed, cookies):
    greed.sort()
    cookies.sort()
    i = j = 0
    while i < len(greed) and j < len(cookies):
        if cookies[j] >= greed[i]:
            i += 1 # chile satified move to next child
        j += 1 # else child is not satiesfied with cookies
    return i
````

````python
g = [1, 2, 3]
s = [1, 1, 3]
match_cookies(g, s)
````

## 🍪 Problem Story — “Feed the Kids”

You have:

* `g[i]`: the **greed factor** of each child → minimum cookie size that satisfies them
* `s[j]`: the **size** of each cookie

You want to give at most **one cookie per child**,
and you want to **maximize the number of happy children**.

---

### 💡 Example

```python
g = [1, 2, 3]     # children's greed
s = [1, 1]        # cookie sizes
```

* Child 1 (greed=1) → gets cookie size 1 ✅
* Child 2 (greed=2) → cookie left is 1 ❌ (too small)

➡️ Only **1 child is happy** → output = `1`

---

## 🧠 Greedy Idea

> Give the **smallest cookie that satisfies** each child.

Why greedy?

* Giving a large cookie to a low-greed child wastes it.
* Giving a small one first leaves bigger cookies for greedier kids.

So:

1. Sort both arrays.
2. Match from smallest child → smallest cookie.
3. If cookie ≥ child’s greed → child is happy → move to next child.

---

## 🧩 Step-by-Step Example

```
g = [1, 2, 3]
s = [1, 1, 3]
```

**Sorted:**
g → [1, 2, 3]
s → [1, 1, 3]

Now pair them:

| Child Greed | Cookie Size | Happy? | Next             |
| ----------- | ----------- | ------ | ---------------- |
| 1           | 1           | ✅      | Move both        |
| 2           | 1           | ❌      | Move cookie only |
| 2           | 3           | ✅      | Move both        |
| 3           | –           | ❌      | Done             |

✅ Total happy kids = 2

## 💻 Code (Short & Clean)

```python
class Solution:
    def findContentChildren(self, g: List[int], s: List[int]) -> int:
        g.sort()
        s.sort()
        i = j = 0
        while i < len(g) and j < len(s):
            if s[j] >= g[i]:
                i += 1  # child satisfied
            j += 1  # move to next cookie
        return i
```

## 🧮 Time Complexity

* Sorting: O(n log n)
* Matching: O(n)
* Total: **O(n log n)**

---

## 🧠 Key Greedy Principle Here

> “Always match the smallest possible resource to the smallest demand.”

Pattern name: **Greedy by sorting and pairing**
Used in:

* Job assignments
* Minimum cost matching
* Interval scheduling
