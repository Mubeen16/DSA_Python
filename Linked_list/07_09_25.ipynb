
## 🧠 Think of `head` like an **entry point**

Imagine a **train** 🚆:

* Each coach is a **Node** (it stores passengers = `data`, and a hook = `next`).
* The **engine** is the **head**.
* If you lose the engine, you can’t pull the train → because you don’t know where it starts.

```
head → [10] → [20] → [30] → None
```

Without `head`, you have no way to find the rest of the nodes.
That’s why every Linked List *must* keep a reference to its **first node**.

---

## 🔎 How it works in code

When we say:

```python
ll.head = node1
```

We’re saying:
👉 “This LinkedList starts at node1.”

Now, if you want to print the list:

```python
current = ll.head   # start at head
while current:      # walk until None
    print(current.data)
    current = current.next
```

The magic is:

* Start from **head** (engine).
* Follow `next` pointers one by one.
* You’ll eventually reach the end (`None`).

---

## ⚡ Why not just use `node1` directly?

Good question!

* In small tests, yes, we can just say `node1 = Node(10)` and start from there.
* But in a **LinkedList class**, we don’t want to expose `node1`, `node2`, etc.
* Instead, we keep **one reference: `head`**.
* That way, no matter how big the list is, you can always start from the beginning and traverse.

---

✅ Summary:

* **Head = the starting point of the list**.
* Without head, the whole list is lost.
* In Python `list`, index `0` is the start.
* In Linked List, `head` is the start.



````python
#creating a head

class Node:
    def __init__(self, data):
        self.data = data
        self.next = None

class LinkedList:
    def __init__(self):
        self.head = None

ll = LinkedList()

node1 = Node(10)
node2 = Node(20)
node3 = Node(30)

ll.head = node1
node1.next = node2
node2.next = node3
````

````python
current = ll.head
while current:
    print(current.data, end=" → ")
    current = current.next
print('None')
````
