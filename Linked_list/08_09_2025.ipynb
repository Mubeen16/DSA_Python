```python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None

class LinkedList:
    def __init__(self):
        self.head = None


    def count_node(self):
        count = 0
        current = self.head
        while current:
            count += 1
            current = current.next
        return count

ll = LinkedList()
node1 = Node(10)
node2 = Node(20)
node3 = Node(40)
node4 = Node(50)
node5 = Node(60)
node6 = Node(70)

ll.head = node1
node1.next = node2
node2.next = node3
node3.next = node4
node4.next = node5
node5.next = node6
```

```python
print(ll.count_node())
```

```python
def insert_at_beginning(self, data):
    new_node = Node(data)        # create node
    new_node.point = self.head   # link new node to old head
    self.head = new_node         # move head to new node
```

```python
def insert_at_end(self, data):
    new_node = Node(data)
    if not self.head:            # if list empty
        self.head = new_node
        return
    current = self.head
    while current.point:         # go to last node
        current = current.point
    current.point = new_node     # attach at end
```

```python
ll = LinkedList()
ll.insert_at_beginning(20)  # List: 20
ll.insert_at_beginning(10)  # List: 10 → 20
ll.insert_at_end(30)        # List: 10 → 20 → 30
ll.insert_at_end(40)        # List: 10 → 20 → 30 → 40

ll.print_nodes()
```
