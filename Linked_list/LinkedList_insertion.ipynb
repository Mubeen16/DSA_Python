head → [2] → [3] → [4] → [5] → None


🧱 Step 1: Create a new node

We create a node newNode with data = 1 and next = None (initially).

newNode → [1 | None]


🔗 Step 2: Point newNode.next to current head

Now we connect newNode.next to the current head, so that the new node points to the old first node.

newNode → [1] ─────┐
                   ↓
           head → [2] → [3] → [4] → [5] → None

🪄 Step 3: Update head to newNode

We now make the new node the head of the linked list.
head → [1] → [2] → [3] → [4] → [5] → None


```python
class Node:
    def __init__(self, data): #we usually have (self, data)
        self.data = data  
        self.next = None

def insertAtFront(head, data):
    newNode = Node(data)
    newNode.next = head
    return newNode

def printList(head):
    while head is not None:
        print(head.data, end="")
        if head.next is not None:
            print(" > ", end="")
        head = head.next
    print()

head = Node(2)
head.next = Node(3)
head.next.next = Node(4)
head.next.next.next =  Node(5)

data = 1
head = insertAtFront(head, data)
printList(head)
        
```
