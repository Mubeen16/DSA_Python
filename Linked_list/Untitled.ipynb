```python
class Node:
    def __init__(self, value):
        self.value = value
        self.next = None


node1 = Node(10)
node2 = Node(20)
node3 = Node(30)
    
```

```python
node1.next = node2
node2.next = node3
node3.next = node1
```

```python
current = node1
while current:
    print(current.value)
    current = current.next
```

## 🔹 Why does it sometimes show up under “Sliding Window”?

* On LeetCode, tags are a bit loose.
* Because this problem uses **two pointers**, some systems tag it similarly to sliding window problems.
* But here, the two pointers aren’t used to create a window — they’re used to move at different speeds around a list.

So:

* **Sliding Window** → usually for substrings/subarrays problems.
* **This problem** → Two Pointers + Linked List (classic cycle detection).

```python
class solution:
    def hascycle(self, head):
        slow = fast = head
        while fast and fast.next:
            slow = slow.next
            fast = fast.next.next
            if slow is fast:
                return True
        return False
```

```python
class Solution:
    def searchInsert(self, nums, target):
        left, right = 0, len(nums)-1
        while left <= right:
            mid = (left + right)//2
            if nums[mid] == target:
                return mid
            elif: nums[mid] < target:
                left = mid +1
            elif:
                right = mid -1
        return left
```
