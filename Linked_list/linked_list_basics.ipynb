

### ✅ **Step-by-step Execution**

#### 1️⃣ Structure of the linked list

```text
head → [10 | *] → [20 | *] → [30 | *] → [40 | None]
```

#### 2️⃣ Function `traverseList(head)`:

* **Base case:** if `head` is `None`, stop recursion.
* **Recursive case:** print current `head.data`, then call `traverseList(head.next)`.

So the calls happen like this:

| Call Stack         | Action                              | Output       |
| ------------------ | ----------------------------------- | ------------ |
| traverseList(10)   | print(10 → ), call traverseList(20) | `10 -> `     |
| traverseList(20)   | print(20 → ), call traverseList(30) | `20 -> `     |
| traverseList(30)   | print(30 → ), call traverseList(40) | `30 -> `     |
| traverseList(40)   | print(40), call traverseList(None)  | `40`         |
| traverseList(None) | base case, return                   | *(new line)* |

---

### 🧩 **Final Output**

```
10 -> 20 -> 30 -> 40
```

---

### 💡 Key Insights

* Each recursive call “remembers” where it was and returns after the deeper call finishes.
* The recursion naturally stops when `head` becomes `None`.
* It’s functionally equivalent to a `while` loop traversal, just using the **call stack** instead of iteration.

## ⚙️ Recursive Function

```python
def traverseList(head):
    if head is None:
        print()
        return
    print(head.data, end="")
    if head.next is not None:
        print(" -> ", end="")
    traverseList(head.next)
```

---

---

## 🔄 Stack Unwinding (Returning)

Once base case returns, each previous call completes and exits:

```
← traverseList(40)
← traverseList(30)
← traverseList(20)
← traverseList(10)
```

Each function finishes one by one until the stack is empty.
