```python
nums = [10, 4, 3, 50, 23, 90]

first = second = float('-inf')

for num in nums:
    if num > first:       # Found new largest
        second = first
        first = num
    elif first > num > second:   # In between first & second
        second = num

print("Second largest:", second)
```

```python
def bubble_sort(arr):
    n = len(arr)
    for i in range(n):
        for k in range(n-i-1):
            if arr[k] > arr[k+1]:
                arr[k], arr[k+1] = arr[k+1], arr[k]
    return arr
```

```python
print(bubble_sort([5, 1, 4, 2, 8]))
```
