```python
class Solutions:
    def getSecondLargest(self, arr):
        first = second = -1

        for x in arr:
            if x > first:
                second = first
                first = x
            elif x > second and x != first:
                second = x
        return second
```

```python
def firstRepeated(arr):
    freq = {}
    for num in arr:
        freq[num] = freq.get(num, 0)+1

    for i, num in enumerate(arr):
        if freq[num] > 1:
            return i + 1
    return -1
        
```

```python


print(firstRepeated([1, 5, 3, 4, 3, 5, 6]))  # 2  (5 repeats first)
print(firstRepeated([1, 2, 3, 4]))           # -1 (no repeats)
print(firstRepeated([7, 7, 8, 9]))           # 1  (7 repeats, first index = 1)
```

```python
class Solution:
    def countZeros(self, arr):
        n = len(arr)
        low , high  = 0, n-1
        first_zero = -1

        while low <= high:
            mid = (low +high)//2

            if arr[mid] == 0:
                first_zero = mid
                high = mid -1
            else:
                low = mid + 1
        if first_zero == -1:
            return 0
        return n - first_zero
```
