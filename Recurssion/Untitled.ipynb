```python
def check_paliindrome(s):
    freq = {}
    n = len(s)
    vow = 'aeiou'
    for i in range(n):
        if s == s[::-1]:
            if s[i] in vow:
                freq += s[i]
            return freq
    return s[::-1]
```

```python
s = "malayalam"

check_paliindrome(s)
```

```python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None

def insert_at_front(head, x):
    new_node = Node(x)
    new_node.next = head        # point new node to old head
    return new_node             # new node becomes head

def insert_at_end(head, x):
    new_node = Node(x)
    if head is None:            # empty list
        return new_node
    curr = head
    while curr.next is not None:
        curr = curr.next
    curr.next = new_node        # link at the tail
    return head                 # head unchanged

def print_list(head):
    curr = head 
    while curr:
        print(curr.data, end='-->' if curr.next else "\n")
        curr = curr.next

head = Node(2)
head.next = Node(3)

head = insert_at_front(head, 1)  # 1 -> 2 -> 3
head = insert_at_end(head, 4)    # 1 -> 2 -> 3 -> 4
print_list(head)
```

```python
def sum_avg(arr):
    sum_arr = sum(arr)
    avg = sum_arr / len(arr)
    return sum_arr, avg
```
