```python
def sum(n):
    if n == 1:
        return 1
    return n + sum(n-1)
```

```python
def fact(n):
    if n == 0:
        return 1
    return n * fact(n-1)
    

fact(5)
```

```python
def printNos(n):
    if n == 0:
        return
    printNos(n-1)
    print(n)
```

```python
n = 4
printNos(n)
```
