Q: given an array of integers and an integer K, find the maximum sum of any contiguous subarray of size K


A **subarray** is always defined as a **continuous slice in the original order of the array**.

👉 So if the array is:
`[2, 1, 5, 1, 3, 2]`

* Subarray of size 3 starting from **left to right**:
  `[2,1,5]`, `[1,5,1]`, `[5,1,3]`, `[1,3,2]`.

* If you scan **right to left**, you’ll get the *same set* of subarrays, just in reverse order.
  `[1,3,2]`, `[5,1,3]`, `[1,5,1]`, `[2,1,5]`.

⚡ **Important:** The definition of subarray does not depend on the direction you look.
It just means:

* take a **starting index i**
* and an **ending index j (i ≤ j)**
* subarray = all elements from `arr[i]` to `arr[j]`.

---

### ✅ So:

* `[2,1,5]` is a subarray.
* `[5,1,2]` (taking non-contiguous jumps) ❌ is **not** a subarray.


````python
def max_sum_subarray(arr, k):
    n = len(arr)
    if n < k:
        return None
    window_sum = sum(arr[:k])
    maxSum = window_sum

    for i in range(k, n):
        window_sum += arr[i] - arr[i-k]
        maxSum = max(maxSum, window_sum)
    return maxSum



print(max_sum_subarray([2,1,5,1,3,2], 3))  # 9
print(max_sum_subarray([1,2,3,4,5], 2))   
````


### Rule Recap

At every step:

```python
window_sum = window_sum - arr[i-k] + arr[i]
```

* `arr[i]` → new element entering from the **right**.
* `arr[i-k]` → old element leaving from the **left**.

---

### Why does `i-k` always point to the “leaving” element?

Because:

* First window covers indices `0..k-1`.
* Loop starts at `i = k` (the first new element).
* The window’s **left boundary** is always `(i-k+1)`.
* So the element at `(i-k)` is just before that boundary → the one leaving.

---

### Walkthrough with k=3

Array = `[2, 1, 5, 1, 3, 2]`

---

**i = 3** (new element `arr[3] = 1`)

* Old window = indices `0..2` → `[2,1,5]`
* Leaving index = `i-k = 3-3 = 0` → `arr[0] = 2`
* Entering index = `i = 3` → `arr[3] = 1`
* New window = indices `1..3` → `[1,5,1]`

✅ Formula: `8 - 2 + 1 = 7`

---

**i = 4** (new element `arr[4] = 3`)

* Old window = indices `1..3` → `[1,5,1]`
* Leaving index = `i-k = 4-3 = 1` → `arr[1] = 1`
* Entering index = `i = 4` → `arr[4] = 3`
* New window = indices `2..4` → `[5,1,3]`

✅ Formula: `7 - 1 + 3 = 9`

---

**i = 5** (new element `arr[5] = 2`)

* Old window = indices `2..4` → `[5,1,3]`
* Leaving index = `i-k = 5-3 = 2` → `arr[2] = 5`
* Entering index = `i = 5` → `arr[5] = 2`
* New window = indices `3..5` → `[1,3,2]`

✅ Formula: `9 - 5 + 2 = 6`

---

### 🚀 The Pattern You Saw

* At **i=3**, we dropped `arr[0]`
* At **i=4**, we dropped `arr[1]`
* At **i=5**, we dropped `arr[2]`

So each time the leaving index increases by +1 (0 → 1 → 2), which matches the **window sliding right**.

That’s why the formula works “automatically” — the math handles shifting.

---

⚡ Behind the scenes:

* `i` walks along the array (new right side).
* `i-k` walks along the array (old left side).
* Together they keep the window size fixed at **k**.
