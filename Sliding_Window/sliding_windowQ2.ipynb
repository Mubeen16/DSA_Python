```python
def average_of_subarrays(arr, k):
    n = len(arr)
    result = []
    window_sum = sum(arr[:k])
    result.append(window_sum/k)

    for i in range(k, n):
        window_sum += arr[i] - arr[i-k]
        result.append(window_sum/k)
    return result


print(average_of_subarrays([1, 3, 2, 6, -1, 4, 1, 8, 2], 5))
# [2.2, 2.8, 2.4, 3.6, 2.8]
```
