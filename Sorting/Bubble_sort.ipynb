
### 🌀 **What is Bubble Sort?**

**Bubble sort** is a simple sorting algorithm that repeatedly **compares adjacent elements** and **swaps them if they are in the wrong order**.
After each full pass, the **largest element “bubbles up”** to the end of the list.

---

### 📘 **Algorithm Logic (Step-by-Step)**

Suppose you have an array:
`[5, 3, 8, 4, 2]`

1. Compare the **first two elements** → 5 and 3 → since 5 > 3, swap them.
   → `[3, 5, 8, 4, 2]`
2. Move to the next pair → 5 and 8 → no swap.
3. Next pair → 8 and 4 → swap.
   → `[3, 5, 4, 8, 2]`
4. Next pair → 8 and 2 → swap.
   → `[3, 5, 4, 2, 8]`

✅ After first pass, **8 is at its correct position** (the end).
Repeat the same process for the remaining elements (excluding the last sorted one each time).

---

### 🧠 **Why It Works**

Each pass places one element in its correct position at the end of the array.
After `n-1` passes, the array becomes fully sorted.

---


### ⚙️ **Explanation of Loops**

* `for i in range(n - 1)`:
  → We need at most `n-1` passes.

* `for j in range(0, n - i - 1)`:
  → After every pass, the last `i` elements are already sorted,
  so we don’t compare them again.

---

### ⏱️ **Time Complexity**

| Case    | Complexity | Explanation                                                |
| ------- | ---------- | ---------------------------------------------------------- |
| Best    | O(n)       | If the array is already sorted (can optimize with a flag). |
| Average | O(n²)      | Typical case — many swaps.                                 |
| Worst   | O(n²)      | Reversed order.                                            |

### 💾 **Space Complexity:** O(1) → in-place sorting (no extra memory).

---

### ⚡ **Optimized Version (with Early Stop)**

```python
def bubble_sort_optimized(arr):
    n = len(arr)
    for i in range(n - 1):
        swapped = False
        for j in range(0, n - i - 1):
            if arr[j] > arr[j + 1]:
                arr[j], arr[j + 1] = arr[j + 1], arr[j]
                swapped = True
        if not swapped:
            break  # stop if no swap happened
    return arr
```




````python

def bubble_sort(arr):
    n = len(arr)
    # Outer loop for passes
    for i in range(n - 1):
        # Inner loop for adjacent comparisons
        for j in range(0, n - i - 1):
            if arr[j] > arr[j + 1]:
                # Swap if elements are in wrong order
                arr[j], arr[j + 1] = arr[j + 1], arr[j]
    return arr

# Example
nums = [5, 3, 8, 4, 2]
print("Before sorting:", nums)
sorted_nums = bubble_sort(nums)
print("After sorting:", sorted_nums)
````
