````python
def selection_sort(arr):
    n = len(arr)
    for i in range(n-1):
        min_index = i
        for j in range(i+1, n):
            if arr[j] < arr[min_index]:
                min_index = j
        arr[i], arr[min_index] = arr[min_index], arr[i]
    return arr
````

````python
nums = [64, 25, 12, 22, 11]
print("Before sorting:", nums)
sorted_nums = selection_sort(nums)
print("After sorting:", sorted_nums)
````



## 🧠 Concept: What is **Selection Sort**?

**Selection Sort** is one of the simplest sorting algorithms.
It works by **repeatedly selecting the smallest element** from the **unsorted part** of the array and placing it at the **beginning**.

---

### 🔁 Step-by-step idea

Suppose we have this list:
`[64, 25, 12, 22, 11]`

We will:

1. Find the smallest element in the entire list → `11`
2. Swap it with the first element → `[11, 25, 12, 22, 64]`
3. Now ignore the first element (it’s sorted) and find the smallest among the rest `[25, 12, 22, 64]` → `12`
4. Swap it with 25 → `[11, 12, 25, 22, 64]`
5. Repeat until the list is sorted.

Final sorted array → `[11, 12, 22, 25, 64]`



---

### 🧮 Dry Run Example

| Step | i | Unsorted Part    | Minimum Found | Swap Result      |
| ---- | - | ---------------- | ------------- | ---------------- |
| 1    | 0 | [64,25,12,22,11] | 11            | [11,25,12,22,64] |
| 2    | 1 | [25,12,22,64]    | 12            | [11,12,25,22,64] |
| 3    | 2 | [25,22,64]       | 22            | [11,12,22,25,64] |
| 4    | 3 | [25,64]          | 25            | [11,12,22,25,64] |

---

### ⏱️ Time Complexity

| Case    | Time  | Explanation               |
| ------- | ----- | ------------------------- |
| Best    | O(n²) | Still compares every pair |
| Average | O(n²) | Typical case              |
| Worst   | O(n²) | Reverse order input       |
| Space   | O(1)  | In-place sorting          |

---

### ✅ Key Notes

* **Not stable** (equal elements can change order)
* **In-place** (no extra memory)
* Good for **small arrays** or **learning purposes**






## 🧩 1️⃣ Why `for i in range(n - 1)` in the **outer loop**

```python
for i in range(n - 1):
```

* `i` marks the **current position** where we’ll put the next smallest element.
* The list has `n` elements — but after you’ve placed `n-1` elements correctly,
  the **last one is automatically sorted** (it must be the largest).

✅ Example:
`arr = [11, 12, 22, 25, 64]`

| When i = | Task                                                          |
| -------- | ------------------------------------------------------------- |
| 0        | Find smallest in full array, put at position 0                |
| 1        | Find smallest in remaining `[25,12,22,64]`, put at position 1 |
| 2        | Find smallest in `[25,22,64]`, put at position 2              |
| 3        | Find smallest in `[25,64]`, put at position 3                 |
| 4        | Nothing left — already sorted! ✅                              |

If you looped to `n`, you'd be doing one **extra, unnecessary** pass.

👉 That’s why we go only up to `n - 1`.

---

## 🧩 2️⃣ Why `for j in range(i + 1, n)` in the **inner loop**

```python
for j in range(i + 1, n):
```

Here’s what’s happening:

* At this point, the subarray **before index i** is already sorted.
* So you only need to look for the **smallest element** in the **unsorted part** —
  which starts right **after** `i`, i.e., from `i + 1`.

---

✅ Example:

Suppose `arr = [11, 25, 12, 22, 64]` and now `i = 1`.

* That means index `0` (`11`) is already sorted.
* The **unsorted** part is `[25, 12, 22, 64]`.
* So we start searching from `j = i + 1 = 2` to `n - 1 = 4`.

| j | arr[j] | Compare with | Action                     |
| - | ------ | ------------ | -------------------------- |
| 2 | 12     | 25           | smaller → update min_index |
| 3 | 22     | 12           | not smaller                |
| 4 | 64     | 12           | not smaller                |

Then we swap 12 and 25.

---

## 🧠 Intuition Recap

| Code Part | Meaning                           | Why                                    |
| --------- | --------------------------------- | -------------------------------------- |
| `n - 1`   | Do n−1 passes                     | Last element automatically sorted      |
| `i + 1`   | Start comparing from next element | Elements before `i` are already sorted |
| `n`       | End boundary                      | Go till last element                   |

---

💬 In short:

> 🔹 Outer loop (`i`) decides *where to place* the next smallest.
> 🔹 Inner loop (`j`) *searches for that smallest* in the remaining list.
