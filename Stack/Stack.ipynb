```python
class Stack:
    def __init__(self):
        self.stack = []

    def push(self, item):
        self.stack.append(item)
    
    def pop(self):
        if not self.is_empty():
            return self.stack.pop()
        else:
            return("stack underflow, cannot pop from emty stack")

    def peek(self):
        if not self.is_empty():
            return self.stack[-1]
        else:
            print("stack is empty")

    def is_empty(self):
        return len(self.stack) == 0

    def display(self):
        print("Stack from top → bottom:", self.stack[::-1])

        
```
