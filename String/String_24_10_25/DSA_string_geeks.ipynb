# 🧩 **Python String Comparison Cheat Sheet**

### 💡 Overview

Python supports several operators to **compare strings**, checking for both **equality** and **alphabetical (lexicographical)** order.

---

## 🟢 **1️⃣ Equality Operators**

### ✅ `==` → Equal

Returns `True` if both strings have identical characters **in same order**.

```python
s1 = "Python"
s2 = "Python"
print(s1 == s2)     # True
```

### 🚫 `!=` → Not Equal

Returns `True` if strings differ.

```python
s1 = "Python"
s2 = "Java"
print(s1 != s2)     # True
```

---

## 🟠 **2️⃣ Lexicographical Comparison (Alphabetical Order)**

Uses Unicode (alphabetical) ordering for `<`, `<=`, `>`, `>=`.

```python
s1 = "apple"
s2 = "banana"

print(s1 < s2)   # True  ('apple' comes before 'banana')
print(s2 > s1)   # True
```

📘 **Tip:** Great for sorting strings alphabetically.

---

## 🔵 **3️⃣ Case-Insensitive Comparison**

Convert both strings to the same case before comparing:

```python
s1 = "Apple"
s2 = "apple"

print(s1.lower() == s2.lower())   # True
print(s1.upper() == s2.upper())   # True
```

---

## 🟣 **4️⃣ Using `startswith()` / `endswith()`**

These check whether a string begins or ends with a given substring.

```python
s = "hello world"
print(s.startswith("hello"))   # True
print(s.endswith("world"))     # True
```

Use in conditional checks or filtering data.

---

## 🧠 **5️⃣ Summary Table**

| Operator / Method        | Meaning                | Case Sensitive | Example                  | Result |
| ------------------------ | ---------------------- | -------------- | ------------------------ | ------ |
| `==`                     | Equal                  | ✅ Yes          | `"hi" == "Hi"`           | False  |
| `!=`                     | Not Equal              | ✅ Yes          | `"hi" != "Hi"`           | True   |
| `<`, `>`, `<=`, `>=`     | Lexicographic order    | ✅ Yes          | `"apple" < "banana"`     | True   |
| `s.lower() == t.lower()` | Case-insensitive equal | ❌ No           | `"Hi" vs "hi"`           | True   |
| `s.startswith("x")`      | Begins with substring  | ✅ Yes          | `"data".startswith("d")` | True   |
| `s.endswith("y")`        | Ends with substring    | ✅ Yes          | `"data".endswith("a")`   | True   |

---

## 🧩 **6️⃣ Real-World Use Cases**

* Sorting a list of names alphabetically
* Comparing user inputs ignoring case
* Filtering data by prefixes (e.g., `"https"`)
* Checking if a file ends with `.csv` or `.py`
