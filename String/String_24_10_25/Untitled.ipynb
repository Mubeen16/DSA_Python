```python
txt = "geeksforgeeks"
pat = "eks"
```

```python
def findsubstring(txt, pat):
    n = len(txt) #getting length for string and substirng
    m = len(pat)

    for i in range(n-m+1): # looop only until the length of substring- minus from string 
        j = 0   #initialising the j
        while j < m and txt[i + j] == pat[j]: 
            j += 1
        if j == m:
            return i
    return -1
```

```python
txt.find(pat)

'''
find() returns:

The index of the first match, or

-1 if not found.

✅ Much faster because Python’s built-in string search uses optimized C-level algorithms internally
'''
```
