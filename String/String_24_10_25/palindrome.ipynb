```python
def is_palindrome(s):
    left = 0  #O(1)
    right = len(s)-1  # O(1)

    while left < right:        # O(n)
        if s[left] != s[right]: 
            return 0
        left += 1 
        right -= 1
    return 1

```

```python
s = "abba"
print(is_palindrome(s))
```

```python
def is_palindrome(s):
    length = len(s)
    for i in range(length//2):
        if s[i] != s[length-i-1]:
            return 0
    return 1
```

```python
s = "abba"
print(is_palindrome(s))
```

```python
def reverseString(s):
    res = []
    for i in range(len(s)-1,-1,-1):
        res.append(s[i])

    return ''.join(res)
    
```

```python
s = "abdcfe"
print(reverseString(s))
```
