## 🧩 Problem Recap

**Goal:**
Given a string `s` that may contain multiple dots (`.`) — including leading, trailing, or consecutive ones — reverse the **order of words**, ensuring:

* Only one dot separates words
* No leading/trailing dots in the final string

---

### 🧠 Example

```
Input:  s = "..geeks..for.geeks."
Output: geeks.for.geeks
```

Explanation → after removing extra dots: words = [`geeks`, `for`, `geeks`], then reversed → `geeks.for.geeks`.

---

## 💡 Approach 1 — Using Stack (O(n) Time + O(n) Space)

```python
def reverseWords(s):
    stack = []
    word = ""

    for ch in s:
        if ch != '.':
            word += ch
        elif word:
            stack.append(word)
            word = ""

    if word:
        stack.append(word)

    return '.'.join(stack[::-1])


# Test
s = "..geeks..for.geeks."
print(reverseWords(s))   # Output: geeks.for.geeks
```

✅ **Logic:**
Push each word to a stack, then pop in reverse order.

---

## ⚡ Approach 2 — Two-Pointer (O(n) Time + O(1) Space)

```python
def reverseWords(s):
    s = s[::-1]          # Reverse entire string
    n = len(s)
    i = 0
    result = []

    l = 0
    while l < n:
        if s[l] != '.':
            if i != 0:
                result.append('.')
                i += 1

            r = l
            while r < n and s[r] != '.':
                result.append(s[r])
                i += 1
                r += 1

            result[i - (r - l):i] = reversed(result[i - (r - l):i])
            l = r
        l += 1
    return ''.join(result)


print(reverseWords("..geeks..for.geeks."))   # geeks.for.geeks
```

✅ **Logic:**
Reverse the whole string → then reverse each word in place.
This minimizes space since it manipulates within O(1).

---

## 🧰 Approach 3 — Pythonic Split & Join (O(n) Time + O(n) Space)

```python
def reverseWords(s):
    words = [word for word in s.split('.') if word]   # Remove empty strings
    return '.'.join(reversed(words))

print(reverseWords("..geeks..for.geeks."))   # geeks.for.geeks
```

✅ **Shortest & Cleanest** using Python built-ins.

---

## 📊 Complexity Summary

| Approach     | Time | Space | Notes                |
| ------------ | ---- | ----- | -------------------- |
| Stack        | O(n) | O(n)  | Step-by-step control |
| Two Pointer  | O(n) | O(1)  | In-place reversal    |
| Split & Join | O(n) | O(n)  | Most concise         |
