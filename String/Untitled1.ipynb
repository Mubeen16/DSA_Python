```python
import numpy as np
A = np.array([[2,  1],
              [1, -1]])

b = np.array([10, 2])
```

```python
print("Solutiona:", np.linalg.solve(A, b))
```

```python
a_inv = np.linalg.inv(A)
```

```python
a_inv
```

```python
sol = a_inv.dot(b)
```

```python
sol
```

```python
A.dot(sol)
```
