```python
def countCharacters(words, chars):
    freq = {}
    for ch in chars:
        freq[ch] = freq.get(ch, 0) + 1

    total_length = 0
    valid_words = []

    for word in words:
        temp = freq.copy()
        can_form = True
        for ch in word:
            if ch not in temp or temp[ch] == 0:
                can_form = False
                break
            temp[ch] -= 1
        if can_form:
            total_length += len(word)
            valid_words.append(word)
    return total_length, valid_words
            
```

```python
# Example
length, words_possible = countCharacters(["cat","bt","hat","tree"], "atach")
print("Total length:", length)         # 6
print("Words that can be formed:", words_possible)  # ['cat', 'hat']
```
