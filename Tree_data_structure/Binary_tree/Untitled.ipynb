binary tree is hierarchiel data structure
each node has most two children=called left and right
a node with no children is called a leaf


```python
# each node stores a value
# a left child
# a right child

class Node:
    def __init__(self, value):
        self.value = value
        self.left = None
        self.right = None


# self.value --> data stored in the node
# self.right and self.left is the pointers to child node

# Create nodes
root = Node(1)
root.left = Node(2)
root.right = Node(3)
root.left.left = Node(4)
root.left.right = Node(5)
```

```python
def inorder(root):
    if root:
        inorder(root.left)
        print(root.value)
        inorder(root.right)
```

```python
inorder(root)
```

```python
def preorder(root):
    if root:
        print(root.value)
        preorder(root.left)
        preorder(root.right)

```

```python
def postorder(root):
    if root:
        postorder(root.left)
        postorder(root.right)
        postorder(root.value)
```
