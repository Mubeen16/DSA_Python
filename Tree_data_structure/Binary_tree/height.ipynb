

## 🌳 Concept

**Height (or depth)** of a binary tree =
👉 the number of **edges** on the longest path from the **root** to a **leaf node**.

Example:

```
      12
     /  \
    8    18
   / \
  5  11
```

Here the longest path is `12 → 8 → 5` (or `12 → 8 → 11`),
which has **2 edges**, so **height = 2**.

---

## 🧩 Approach 1 — Recursive (Top-Down or Bottom-Up)

```python
class Node:
    def __init__(self, val):
        self.data = val
        self.left = None
        self.right = None

def height(root):
    if root is None:
        return -1   # returning -1 because height = number of edges

    lHeight = height(root.left)
    rHeight = height(root.right)

    return max(lHeight, rHeight) + 1
```

### 🧠 Logic:

* For each node, compute height of left & right subtree.
* Height of that node = `1 + max(leftHeight, rightHeight)`.
* Base case: if no node → height = `-1` (so a single node tree’s height = 0 edges).

### 🧮 Complexity:

* **Time:** O(n) → every node visited once.
* **Space:** O(n) → recursive stack.

## 🧱 Approach 2 — Level Order Traversal (BFS)

```python
from collections import deque

class Node:
    def __init__(self, data):
        self.data = data
        self.left = None
        self.right = None

def height(root):
    if root is None:
        return 0

    q = deque([root])
    depth = 0

    while q:
        levelSize = len(q)

        for _ in range(levelSize):
            curr = q.popleft()

            if curr.left:
                q.append(curr.left)
            if curr.right:
                q.append(curr.right)

        depth += 1

    return depth - 1
```

### 🧠 Logic:

* Traverse level by level.
* For each level processed, increment `depth`.
* When queue becomes empty, `depth - 1` gives the number of edges.

### 🧮 Complexity:

* **Time:** O(n)
* **Space:** O(n) (queue holds one level’s nodes at a time)
