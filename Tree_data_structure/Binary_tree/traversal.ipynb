```python
class Node:
    def __init__(self, data):
        self.data = data
        self.left = None
        self.right = None

root = Node(2)
root.left = Node(3)
root.right = Node(4)
root.left.left = Node(5)
```

```python
def inorder(root):
    if root:
        inorder(root.left)
        print(root.data, end=" ")
        inorder(root.right)
```

```python
def preorder(root):
    if root:
        print(root.data)
        preorder(root.left)
        preorder(root.right)
        
```

```python
def postorder(root):
    if root:
        postorder(root.left)
        postorder(root.right)
        print(root.data)
```

Level Order Traversal is a way of visiting all the nodes of a binary tree level by level — that is, from top to bottom and left to right across each level.

        1
      /  \
     2    3
    / \   /\
   4  5  6  7


```python
from collections import deque

class Node:
    def __init__(self, value):
        self.data = value
        self.left = None
        self.right = None

def levelOrder(root):
    if not root:
        return []

    q = deque([root])
    res = []

    while q:
        level_nodes = []
        for _ in range(len(q)):      # process one level at a time
            node = q.popleft()
            level_nodes.append(node.data)

            if node.left:
                q.append(node.left)
            if node.right:
                q.append(node.right)

        res.append(level_nodes)

    return res
```
