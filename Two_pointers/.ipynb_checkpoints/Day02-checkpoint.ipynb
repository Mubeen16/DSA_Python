````python
def is_palindrome(s):
    cleaned = ''
    for char in s:
        if char.isalnum():
            cleaned += char
    return cleaned == cleaned[::-1]
            
````

````python
s = "a 1 b 2 b 1 a"
print(is_palindrome(s))
````


# **Valid Palindrome — Two Pointers Approach**

## **Problem**

Check if a string is a palindrome **after**:

* Removing all non-alphanumeric characters.
* Ignoring letter case.

---

## **Algorithm (Step-by-step)**

### **1. Initialize pointers**

```python
left = 0
right = len(s) - 1
```

* `left` → start of the string
* `right` → end of the string

---

### **2. Loop until pointers meet**

We use:

```python
while left < right:
```

* Continue checking until `left` and `right` meet or cross.

---

### **3. Skip non-alphanumeric characters**

```python
if not s[left].isalnum():
    left += 1
    continue

if not s[right].isalnum():
    right -= 1
    continue
```

* `.isalnum()` checks if a character is **a letter (A–Z, a–z)** or **a digit (0–9)**.
* `not s[left].isalnum()` means the character is **NOT** a letter/digit (it’s punctuation, space, etc.).
* If found → move the pointer inward (`left += 1` or `right -= 1`) and **skip the rest of the loop** for this iteration (`continue`).
* Do this for **both** left and right pointers.

---

### **4. Compare the characters**

```python
if s[left].lower() != s[right].lower():
    return False
```

* Convert both characters to lowercase for **case-insensitive** comparison.
* If they **don’t** match → the string is **not** a palindrome → return `False` immediately.

---

### **5. Move inward if matched**

```python
left += 1
right -= 1
```

* Only move after a successful match.
* This shrinks the checking range.

---

### **6. End condition**

When `left >= right`, we’ve checked everything:

```python
return True
```

* If no mismatches found → it’s a palindrome.

---

## **Full Code**

```python
def is_palindrome_two_pointers(s):
    left, right = 0, len(s) - 1

    while left < right:
        # Skip non-alphanumeric from left
        if not s[left].isalnum():
            left += 1
            continue

        # Skip non-alphanumeric from right
        if not s[right].isalnum():
            right -= 1
            continue

        # Compare lowercase characters
        if s[left].lower() != s[right].lower():
            return False

        # Move both pointers inward
        left += 1
        right -= 1

    return True
```

---

## **Example Walkthrough**

`s = "A man, a plan, a canal: Panama"`

* **Step 1:** left=`A`, right=`a` → match
* **Step 2:** left=` ` (space) → skip
* **Step 3:** left=`m`, right=`m` → match
* … repeat skipping and comparing …
* All matches → return `True`.

---

## **Complexity**

* **Time:** O(n) — one pass through the string.
* **Space:** O(1) — only two pointers, no extra strings.





````python
class Solution:

    def isPalindrome(self, s: str) -> bool:

        string = [c for c in s.lower() if c.isalnum()]

        if len(string)==0:
            return True

        return string == list(reversed(string))
````

````python
def sorted_squares(nums):
    n = len(nums)
    left = 0
    right = n -1
    result = [0]*n
    pos = n-1
    while left <= right:
        if abs(nums[left]) > abs(nums[right]):
            result[pos] = nums[left]**2
            left += 1
        else:
            result[pos] = nums[right]**2
            right -= 1
        pos -= 1
    return result 
            
            
````

````python
nums = [-4, -1, 0, 3, 10]
print(sorted_squares(nums))
````
