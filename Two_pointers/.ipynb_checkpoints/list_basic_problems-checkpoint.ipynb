````python
def swapList(list):
    start, *middle, end = list
    list = [end, *middle, start]
    return list

newList = [12, 35, 9, 56, 24]

print(swapList(newList))
````



> "How do I know when I should put something inside the `()` of a method, and when to leave it empty — like why `.append()` takes a value but `.reverse()` doesn’t?"

---

## **The rule is simple:**

* If the method **needs extra information from you** → you must pass it inside `()`.
* If the method **already knows what to do just from the object itself** → `()` stays empty.

---

### **1️⃣ Methods that need extra info → you pass it**

These methods **cannot do their job** without knowing *what* to act on.

Example:

```python
a = [1, 2, 3]
a.append(4)         # Needs the value to append
a.insert(1, 99)     # Needs both position and value
a.remove(2)         # Needs the value to remove
a.index(3)          # Needs the value to search
```

💡 Without the extra info, Python wouldn’t know *what* you want.

---

### **2️⃣ Methods that already know what to do → leave empty**

These methods **work only with the object itself**, so they don’t need more info.

Example:

```python
a = [1, 2, 3]
a.reverse()   # Just flips the list in place — no extra input needed
a.sort()      # Sorts the list in place — uses list content
a.clear()     # Removes all elements
```

💡 These methods are **self-contained** — the object gives them all they need.

---

### **Why `.append()` and `.reverse()` are different**

* `.append()` → needs to know *what* to add → argument required.
* `.reverse()` → just flips what’s already there → no argument.

---

✅ **Mental shortcut**:

* **Ask yourself:** “Could Python guess what I want without me telling it anything?”

  * If **no** → put something in `()`.
  * If **yes** → leave `()` empty.

---

````python
a = [8, 3, 5, 1, 9, 12]

# Initialize "smallest" value with first element of list
smallest = a[0]

# Iterate through list to find smallest element
for val in a:
  
    # If current value is smaller than current smallest value
    if val < smallest:
      
        # Update the smallest value
        smallest = val

print(smallest)
````

````python
# Creating a list of numbers
a = [1, 2, 3, 4, 5, 6, 7, 8, 9]

# Initializing counters for even and odd numbers
even = 0
odd = 0

for num in a:
    if num % 2 == 0:
        even += 1
    else:
        odd += 1
print(even)
print(odd)
````

````python
a = [1, 2, 3, 1, 2, 4, 5, 6, 5]

# Initialize a dictionary to count occurrences
d = {}

for n in a:
    d[n] = d.get(n, 0)+1

dup = [n for n, c in d.items() if c > 1]

print(dup)
````
