# Array: Decalare, indexing, Traversal

# P1.Sum of Array

```python
def arr_sum(a):
    s = 0
    for x in a:   #Traversal
        s += x
    return s
```

# P2. Max & Min (Handle empy)

```python
def arr_max(a):
    if not a: return None
    m = a[0]
    for x in a[1:]:
        if x > m: m = x
    return m

def arr_min(a):
    if not a: return None
    m = a[0]
    for x in a[1:]:
        if x < m: m = x
    return m
    
```

# P3. Count occurance of a value


```python
def count_occ(a, val):
    c = 0
    for x in a:
        if x == val:
            c += 1
    return c
```

# P4. Reverse in-place (two pointers)

```python
def reverse_in_place(a):
    i, j = 0, len(a)-1
    while i < j:
        a[i],a[j] = a[j], a[i]
        i += 1; j -= 1
    return a
```

# P5. Check if sorted non-decreasing

```python
def is_sorted(a):
    for i in range(1, len(a)):
        if a[i] < a[i-1]:
            return False
        return True
```

# P7. Second largest (distinct). Return None if not enough distinct.

```python
def second_largest(a):
    first = second = None
    for x in a:
        if first is None or x > first:
            second = first
            first = x
        elif x != first and (second is None or x > second):
            second = x
    return second
```

Here’s the **short logic** of `second_largest`:

1. Keep two trackers: `first` = largest so far, `second` = second largest so far.
2. For each element `x` in the array:

   * If `x` is bigger than `first` (or `first` is empty), update:
     → `second = first`, `first = x`.
   * Else if `x` is not equal to `first` **and** (we don’t yet have `second` or `x > second`), update:
     → `second = x`.
3. At the end, return `second`.

Maintain top 2 distinct numbers while scanning once.
